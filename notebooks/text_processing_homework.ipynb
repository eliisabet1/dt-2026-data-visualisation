The text I chose is *Alice's Adventures in Wonderland* by Lewis Carroll.



# Task 1 – Read Text and Work With It

For this task, I chose Alice's Adventures in Wonderland by Lewis Carroll.

I will use urllib.request to open the online text file and read it. Then I will use different string operations to work with the text.


```python

import urllib.request

url = "https://www.gutenberg.org/files/11/11-0.txt"

response = urllib.request.urlopen(url)

text = response.read().decode("utf-8")

print(len(text), "characters loaded")
```

## Looking at the text

The text is now stored in a variable called text.

I use slicing to print only the beginning of the text instead of printing the entire book.

```python
print(text[:500])
```

## Working with strings
lower() makes all letters lowercase.

split() separates the text into individual words.

len(words) counts how many words there are.

words[:20] shows the first 20 words.

```python

text = text.lower()

words = text.split()

print("Number of words:", len(words))

print("First 20 words:", words[:20])
```


## Using a for loop

A for loop allows us to go through each word in a list.

Here I print the first ten words from the book, one by one.

```python

for word in words[:10]:
    print(word)
```

## Looping through letters

Python can also go through individual letters in a word.

Here I use a for loop to print each letter of the word Alice.

```python
for letter in "alice":
    print(letter)
```


## Working with indexing and slicing

Indexing allows us to select one character from a string.

Python starts counting from 0. Slicing allows us to select a part of a string.


```python

title = "Alice's Adventures in Wonderland"

print("First character:", title[0])

print("Last character:", title[-1])

print("First five characters:", title[:5])
```

# TASK 2 - Creating a basic world cloud

```python

from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
```

```python
stopwords = set(STOPWORDS)

stopwords.update(["gutenberg", "ebook", "chapter", "illustration"])
```

## Creating the word cloud

I use WordCloud() to generate a word cloud from the text of Alice's Adventures in Wonderland.

The most frequently used words will appear larger than the other words.

```python
wc = WordCloud(
    width=800,
    height=400,
    background_color="white",
    max_words=100,
    stopwords=stopwords
).generate(text)

plt.imshow(wc)
plt.axis("off")
plt.show()
```

## Word Cloud Interpretation

The word cloud shows the most frequently used words in Alice's Adventures in Wonderland. The largest words are "Alice" and "said", which means they appear very often in the story.

Other noticeable words include "queen", "hatter", "began", and "thing". These words give an idea of the characters and events in the story.

I chose a white background and allowed up to 100 words to appear in the word cloud. I also used stopwords to remove some common words and make the word cloud easier to understand.
