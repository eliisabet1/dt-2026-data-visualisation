```python
import pandas as pd 
df = pd.read_csv("Shakespeare.csv")
df
```

```python
genre_counts = df["Genre"].value_counts()
plt.pie( genre_counts, labels=genre_counts.index, autopct="%1.1f%%",  colors=["#FF00C5","#FF69E2", "#FFB0EF"])
plt.title("Shakespeare Plays by Genre")
plt.show()
```

The chart shows the number of Shakespeare plays in each genre. Comedy has the most plays in the dataset, with 40% of plays, followed by history with 36% of plays and tragedy with 24% of plays.

```python
import matplotlib.pyplot as plt
counts = df.groupby(["Approx. Year", "Genre"]).size().reset_index(name="Count")
plt.scatter(counts["Approx. Year"], counts["Genre"], s=counts["Count"] * 400, alpha=0.5, color="#FFC0CB")
plt.title("Shakespeare Plays by Year and Genre")
plt.xlabel("Approx. Year")
plt.ylabel("Genre")
plt.show()
```

The chart shows the number of Shakespeare plays written in each year and their genre. It shows that Shakespeare wrote comedies, histories, and tragedies between around 1590 and 1606. The chart also shows that some years had more plays than others.

```python
import matplotlib.pyplot as plt
setting_counts = df["Main Setting"].value_counts()
setting_counts.plot( kind="bar", color="#6E0063", legend=False)
plt.title("Number of Plays by Main Setting")
plt.xlabel("Main Setting")
plt.ylabel("Number of Plays")
plt.show()
```

The chart shows that England is the most common setting, with 9 plays taking place there. This is mainly because several of Shakespeare’s history plays are set in England and focus on English kings and historical events. Other settings, such as Venice and Verona, appear in fewer plays.

```python
year_counts = df["Approx. Year"].value_counts().sort_index()
year_counts.plot(kind="line", marker="o", color="#FF36D7")
plt.title("Number of Shakespeare Plays by Year")
plt.xlabel("Approx. Year")
plt.ylabel("Number of Plays")
plt.show()
```

The chart shows that 1594 had the most plays, with 3 plays in total. A few other years, including 1591, 1596, 1597, 1604, and 1606, had 2 plays each. Most of the other years had just 1 play.
