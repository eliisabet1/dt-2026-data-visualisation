Create a chart showing who puts in the most effort. effort can be found from nr of hours 

```python
import pandas as pd
df = pd.read_csv("competencies.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
avg_hours= df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", color="#FFC0CB")
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Grade"]*5, alpha=0.4, color="#FFC0CB")
plt.title("hours studied per grade")
plt.ylabel("grade")
plt.xlabel("hours studied")
```

```python
plt.Figure(figsize=(0,5))
df[["Hours_Studied", "Grade"]]. boxplot()
plt.title("box plot for hours studied and grades")
plt.show
```
